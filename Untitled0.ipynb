```python
import pandas as pd
df = pd.read_csv('/content/Superstore.csv', encoding='latin1')
df.head()
```

```python
# Basic info about the dataset
df.info()
print("\nMissing values:\n", df.isnull().sum())
print("\nTotal rows:", len(df))
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')

top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
print(top_products)
```

```python
region_profit = df.groupby('Region')[['Sales', 'Profit']].sum().sort_values(by='Profit', ascending=False)
print(region_profit)
```

```python
import matplotlib.pyplot as plt

region_profit[['Sales', 'Profit']].plot(kind='bar', figsize=(8,5), color=['#4C72B0','#55A868'])
plt.title('Region-wise Sales vs Profit')
plt.ylabel('Amount')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('region_chart.png')
plt.show()
```

```python
top_products.plot(kind='barh', figsize=(8,6), color='#C44E52')
plt.title('Top 10 Products by Sales')
plt.xlabel('Sales')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('top_products_chart.png')
plt.show()
```
